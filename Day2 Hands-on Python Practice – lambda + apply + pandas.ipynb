```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'num':[1,2,3,4,5]})
```

```python
df
```

```python
#Exercise 1: Square numbers in a column

df['sqaured']  = df['num'].apply(lambda x : x*x)
df
```

```python
#Exercise 2: Even or Odd
df['even_odd'] = df['num'].apply(lambda x: "even" if x% 2 == 0 else "odd")
df
```

```python
# Exercise 3: Lowercase names
df['lowercase']  = df['even_odd'].apply(lambda x : x.lower())
df
```

```python
#Exercise 4: Sum of marks

df_marks = pd.DataFrame({
    'math':[90,80,70],
    'science':[85,75,95],
    'english':[88,92,78]})
```

```python
df_marks['total'] = df_marks.apply(lambda x : x['math'] + x['science'] + x['english'], axis = 1)
df_marks
```

```python
#Exercise 5: Best subject

df_marks['best_subject'] = df_marks.apply(lambda x : x.idxmax(), axis = 1)
df_marks  #################################################
```

```python
#Exercise 6: Pass/Fail based on total
df_marks['pass_fail'] = df_marks['total'].apply(lambda x : 'pass' if x > 250 else 'fail')
df_marks
```

```python
# Exercise 7: Conditional bonus marks

df_marks['bonusadded']  = df_marks['total'].apply(lambda x : x + 5 if x < 250 else x)
df_marks
```

```python
#  3: String Manipulation (Medium)
```

```python
df_people = pd.DataFrame({'fullname':['Alice Johnson','Bob Smith','Charlie Brown']})
# extract first name

df_people['first'] = df_people['fullname'].apply(lambda x : x.split()[0])

df_people['last'] = df_people['fullname'].apply(lambda x : x.split()[1])
df_people
```

```python
# Exercise 10: Count number of letters in first name
df_people['letter_first']=  df_people['first'].apply(lambda x : len(x))
df_people
```

```python
#Exercise 11: Apply discount to price
```

```python
df = pd.DataFrame({
    'product':['A','B','C','D'],
    'price':[120,80,150,60]
})
```

```python
df['discounted price'] = df['price'].apply(lambda x : x - (x*0.01) if x > 100 else x)
df
```

```python
# Exercise 12: Categorize price

df['price category'] = df['price'].apply(lambda x : "high" if x > 100 else "low")
df
```

```python
# Exercise 13: Add tax 5% for all products
df['taxadded'] = df['price'].apply(lambda x : x + x*(0.05))
df
```

```python
# Part 5: Date Manipulation (Medium-Hard)
import datetime
```

```python
df = pd.DataFrame({'date':['2026-02-07','2026-02-08','2026-02-09']})
```

```python
# Exercise 14: Convert string to datetime
df['date']  = pd.to_datetime(df['date'])
```

```python
# Exercise 16: Extract weekday name
df['weekdayname']  = df['date'].dt.day_name()
df
```

```python
# Exercise 17: Apply discount based on product category
```

```python
df = pd.DataFrame({
    'product':['A','B','C','D'],
    'price':[120,80,150,60],
    'category':['Electronics','Clothes','Electronics','Toys']
})
```

```python
df['discountedprice'] = df.apply(lambda row: row['price'] *0.8 if row['category'] == 'Electronics' else row['price'], axis = 1)
df  # row wise operation
```
